```python
import pandas as pd
import glob
import os
from pathlib import Path
```

```python
# ==============================
# 1. Define paths
# ==============================
input_path = r'D:\RESEARCH\Youth Tobacco Survey\data merging\europe'
output_path = Path(r'D:\RESEARCH\Youth Tobacco Survey\data merging\all region')

# ==============================
# 2. Get all CSV files
# ==============================
files = glob.glob(os.path.join(input_path, '*.csv'))

# ==============================
# 3. Read + Process + Store
# ==============================
df_list = []

for file in files:
    df = pd.read_csv(file)
    
    # Extract filename
    filename = os.path.basename(file).replace('.csv', '')
    
    # Split country & year
    parts = filename.split('_')
    year = parts[-1]
    country = '_'.join(parts[:-1])
    
    # Fix short year (e.g., 13 → 2013)
    if len(year) == 2:
        year = '20' + year
    
    # Add new columns
    df['Country'] = country
    df['Year'] = int(year)
    df['Region'] = 'Europe'
    
    # Append
    df_list.append(df)

# ==============================
# 4. Combine all files
# ==============================
final_df = pd.concat(df_list, ignore_index=True)

# ==============================
# 5. Save output
# ==============================
final_df.to_csv(output_path / 'Europe.csv', index=False)

print("✅ All files merged successfully!")
```
